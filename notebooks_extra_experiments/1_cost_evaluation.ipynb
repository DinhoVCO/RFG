## Plotting the cost vs nDCG@10 of the different approaches

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.lines import Line2D

# 1. Definir los datos (sin cambios)
all_data_rows = [
    # nfcorpus
    {'Strategy': 'hyde', 'Dataset': 'nfcorpus', 'input': 27.33, 'output': 1713.82, 'nDCG10': 38.1},
    {'Strategy': 'q2d', 'Dataset': 'nfcorpus', 'input': 1713.82, 'output': 444.05, 'nDCG10': 30.8},
    {'Strategy': 'Gen Doc', 'Dataset': 'nfcorpus', 'input': 2166.86, 'output': 357.7, 'nDCG10': 39.9},
    {'Strategy': 'Gen Ans', 'Dataset': 'nfcorpus', 'input': 2166.86, 'output': 177.03, 'nDCG10': 40.1},
    {'Strategy': 'Gen Doc+Gen Ans', 'Dataset': 'nfcorpus', 'input': 4333.72, 'output': 534.73, 'nDCG10': 40.6},
    {'Strategy': 'Original Query', 'Dataset': 'nfcorpus', 'input': 0, 'output': 0, 'nDCG10': 36.1},
    {'Strategy': 'RFG(Query+Doc+Ans)', 'Dataset': 'nfcorpus', 'input': 4333.72, 'output': 534.73, 'nDCG10': 39.8},
    
    # fiqa
    {'Strategy': 'hyde', 'Dataset': 'fiqa', 'input': 35.67, 'output': 2112.8, 'nDCG10': 45.9},
    {'Strategy': 'q2d', 'Dataset': 'fiqa', 'input': 1197.96, 'output': 461.42, 'nDCG10': 43.6},
    {'Strategy': 'Gen Doc', 'Dataset': 'fiqa', 'input': 1294.59, 'output': 381.33, 'nDCG10': 44.5},
    {'Strategy': 'Gen Ans', 'Dataset': 'fiqa', 'input': 1294.59, 'output': 140.91, 'nDCG10': 44.9},
    {'Strategy': 'Gen Doc+Gen Ans', 'Dataset': 'fiqa', 'input': 2589.18, 'output': 522.24, 'nDCG10': 45.8},
    {'Strategy': 'Original Query', 'Dataset': 'fiqa', 'input': 0, 'output': 0, 'nDCG10': 44.3},
    {'Strategy': 'RFG(Query+Doc+Ans)', 'Dataset': 'fiqa', 'input': 2589.18, 'output': 522.24, 'nDCG10': 47.0},

    # scidocs
    {'Strategy': 'hyde', 'Dataset': 'scidocs', 'input': 36.46, 'output': 2341.892, 'nDCG10': 22.8},
    {'Strategy': 'Gen Doc', 'Dataset': 'scidocs', 'input': 1333.719, 'output': 286.376, 'nDCG10': 22.8},
    {'Strategy': 'Gen Ans', 'Dataset': 'scidocs', 'input': 1333.719, 'output': 187.385, 'nDCG10': 23.1},
    {'Strategy': 'Gen Doc+Gen Ans', 'Dataset': 'scidocs', 'input': 2667.438, 'output': 455.761, 'nDCG10': 22.9},
    {'Strategy': 'Original Query', 'Dataset': 'scidocs', 'input': 0, 'output': 0, 'nDCG10': 22.5},
    {'Strategy': 'RFG(Query+Doc+Ans)', 'Dataset': 'scidocs', 'input': 2667.438, 'output': 455.761, 'nDCG10': 23.5},

    # arguana
    {'Strategy': 'hyde', 'Dataset': 'arguana', 'input': 290.8193, 'output': 2100.33, 'nDCG10': 66.1},
    {'Strategy': 'Gen Doc', 'Dataset': 'arguana', 'input': 1559.42460, 'output': 185.6963, 'nDCG10': 63.4},
    {'Strategy': 'Gen Ans', 'Dataset': 'arguana', 'input': 1559.42460, 'output': 118.98790, 'nDCG10': 60.6},
    {'Strategy': 'Gen Doc+Gen Ans', 'Dataset': 'arguana', 'input': 3118.8492, 'output': 304.6842, 'nDCG10': 64.9},
    {'Strategy': 'Original Query', 'Dataset': 'arguana', 'input': 0, 'output': 0, 'nDCG10': 70.3},
    {'Strategy': 'RFG(Query+Doc+Ans)', 'Dataset': 'arguana', 'input': 3118.8492, 'output': 304.6842, 'nDCG10': 68.6},
]

# 2. Preparación de datos y mapas
df = pd.DataFrame(all_data_rows)
df['cost_per_1k'] = ((df['input'] * 0.1) + (df['output'] * 0.3)) / 1000

datasets_to_plot = ['nfcorpus', 'fiqa', 'scidocs', 'arguana']
strategy_order = sorted(df['Strategy'].unique())

strategy_marker_map = {
    'Original Query': 'X', 'hyde': 'o', 'q2d': 's', 'Gen Doc': '^',
    'Gen Ans': 'D', 'Gen Doc+Gen Ans': 'P', 'RFG(Query+Doc+Ans)': '*'
}
color_palette = sns.color_palette("tab10", n_colors=len(strategy_order))
strategy_color_map = dict(zip(strategy_order, color_palette))

baseline_ndcg = {
    'nfcorpus': 36.1, 'fiqa': 44.3, 'scidocs': 22.5, 'arguana': 70.3
}

# 3. Creación del gráfico
plt.style.use('seaborn-v0_8-whitegrid')
# El parámetro sharex=True se elimina para que cada gráfico pueda tener su propio eje X si es necesario
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 13)) 

# Loop para graficar
for i, dataset_name in enumerate(datasets_to_plot):
    ax = axes.flatten()[i]
    df_subset = df[df['Dataset'] == dataset_name]
    base_value = baseline_ndcg[dataset_name]
    ax.axhline(y=base_value, color='red', linestyle='--', linewidth=2)
    
    sns.scatterplot(
        data=df_subset, x='cost_per_1k', y='nDCG10',
        hue='Strategy', style='Strategy',
        hue_order=strategy_order, palette=strategy_color_map,
        markers=strategy_marker_map, s=200, ax=ax,
        legend=False
    )

    for _, row in df_subset.iterrows():
        ax.text(row['cost_per_1k'] + 0.01, row['nDCG10'], row['Strategy'], fontsize=14, verticalalignment='center')
    
    ax.set_title(f'Dataset: {dataset_name.upper()}', fontsize=18, weight='bold')

    # --- CAMBIO: Poner etiquetas en todos los subgráficos ---
    # Se eliminan las condiciones 'if' para que las etiquetas aparezcan siempre.
    ax.set_ylabel('nDCG@10', fontsize=14, weight='bold')
    ax.set_xlabel('Cost per 1,000 Runs ($)', fontsize=14, weight='bold')


# 4. Leyenda Global (sin cambios)
handles = [
    Line2D([0], [0],
           marker=strategy_marker_map[strategy],
           color='w',
           label=strategy,
           markerfacecolor=strategy_color_map[strategy],
           markersize=10)
    for strategy in strategy_order
]

baseline_handle = Line2D([0], [0], color='red', linestyle='--', linewidth=2, label='Baseline nDCG@10')
handles.append(baseline_handle)

fig.legend(
    handles=handles,
    title='Strategy & Baseline',
    loc='lower center',
    bbox_to_anchor=(0.5, 0.01),
    ncol=8,
    fontsize=14,
    title_fontsize=14
)

plt.tight_layout(rect=[0, 0.08, 1, 0.95])
plt.savefig('final_performance_vs_cost_v2.pdf', bbox_inches='tight')
plt.show()
```
