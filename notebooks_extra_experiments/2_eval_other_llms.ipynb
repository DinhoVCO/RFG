# Evaluating Other Generative Models for RFG

For this evaluation, we used LiteLLM to access different model APIs through the OpenAI endpoint.

Only the nfcorpus dataset and the BGE Large embedding were used.


## Imports

```python
from kdir_src.generators.open_ai import OpenAIGenerator
from kdir_src.generators.lite_llm import LiteLLMGenerator
from kdir_src.generators.mistral_ai import MistralGenerator
from kdir_src.models.all_models import load_models
from kdir_src.fire.Fire import Fire
from kdir_src.fire.promptor_fire import Promptor_fire
```

```python
dataset_name='nfcorpus'
promptor = Promptor_fire(dataset_name)
```

## GPT 3.5

```python
generator = OpenAIGenerator('gpt-35-turbo-petrobras', 1, 512, 0.0)
fire_rag = Fire(dataset_name, promptor, generator, None)
fire_rag.generate_openai_doc_and_save_prf('gpt-35','../input_prompt/fire/input_prompt.json','../doc_gen/fire/openai/')
```

## GPT-4o

```python
generator = OpenAIGenerator('gpt-4o-petrobras', 1, 512, 0.0)
fire_rag = Fire(dataset_name, promptor, generator, None)
fire_rag.generate_openai_doc_and_save_prf('gpt-4o','../input_prompt/fire/input_prompt.json','../doc_gen/fire/openai/')
```

## Mistral Large

```python
encoder_models = load_models()
KEY = ''
generator = MistralGenerator('mistral-large-2411', KEY, 1, 512, 0.0)
fire_rag = Fire(dataset_name, promptor, generator, encoder_models)
fire_rag.generate_and_save_prf_bge_large(path='../doc_gen/fire/mistral_large/', batch_size=128, top_k=5)
```

## Mistral medium

```python
encoder_models = load_models()
KEY = ''
generator = MistralGenerator('mistral-medium-2505', KEY, 1, 512, 0.0)
fire_rag = Fire(dataset_name, promptor, generator, encoder_models)
fire_rag.generate_and_save_prf_bge_large(path='../doc_gen/fire/medium/', batch_size=128, top_k=5)
```

## Deepseek-r1-v1

```python
model='deepseek-r1-v1'
generator = LiteLLMGenerator(model, 1, 512, 0.0)
fire_rag = Fire(dataset_name, promptor, generator, None)
fire_rag.generate_openai_doc_and_save_prf(model,'../input_prompt/fire/input_prompt.json',f'../doc_gen/fire/{model}/')
```
